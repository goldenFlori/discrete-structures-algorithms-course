## Algoritmet që trajtohen

Në këtë ushtrim do të shqyrtohen dhe do të implementohen disa nga algoritmet themelore të analizës së grafeve, të cilët shërbejnë si për eksplorimin ashtu edhe për optimizimin.

| Algoritmi | Funksioni kryesor |
|-----------|------------------|
| **BFS (Breadth-First Search)** | Gjen distancën më të shkurtër nga një nyje burim në grafet pa peshë |
| **DFS (Depth-First Search)** | Eksploron grafin, verifikon arritshmërinë dhe zbulon ciklet |
| **Kruskal / Prim** | Ndërtojnë pemën përfshirëse minimum në grafet me peshe |

## Ushtrimi 1 — Zgjidhja e labirintit me BFS dhe DFS

### Teoria: përshkimi i grafeve

Një **labirint** përbën një shembull tipik të një grafi të nënkuptuar: çdo qelizë e hapur shërben si kulm, ndërsa dy qeliza konsiderohen të lidhura me një brinjë nëse është e mundur të kalohet drejtpërdrejt nga njëra te tjetra. Algoritmet BFS dhe DFS janë dy mënyrat kryesore për të përshkuar një graf dhe për të gjetur një rrugë nga një kulm i caktuar në një tjetër.

---

### BFS (Breadth-First Search)

**Ideja kryesore:** shqyrtohen fillimisht të gjithë fqinjët e kulmit aktual, përpara se të kalohet në nivelin e ardhshëm të thellësisë.  
Mund të përfytyrohet si përhapja e një valë uji — ajo zgjerohet njëkohësisht në të gjitha drejtimet, rreth pas rrethi.

**Struktura e të dhënave:** radha — Queue (FIFO: i pari që hyn, i pari që del)

**Hapat e algoritmit:**
1. Vendoset kulmi fillestar në radhë dhe shënohet si i vizituar.
2. Nxirret kulmi nga fillimi i radhës dhe përpunohet.
3. Shtohen në radhë të gjithë fqinjët e pavizituar dhe shënohen si të tillë.
4. Veprimi përsëritet derisa radha të zbrazet ose të arrihet kulmi i kërkuar.

**Veti kryesore:** BFS gjen gjithmonë **rrugën më të shkurtër** (me numrin më të vogël të brinjëve) në grafet pa peshë.

```
Kompleksiteti kohor:  O(V + E)   ku V = numri i kulmeve, E = numri i brinjëve
```

---

### DFS (Depth-First Search)

**Ideja kryesore:** ndiqet një rrugë sa më thellë të jetë e mundur, përpara se të kthehemi pas për të provuar një tjetër.  
Mund të përfytyrohet si eksplorimi i një shpelle — fillimisht ndiqet një tunel deri në fund dhe vetëm pas kësaj kalohet në një tjetër.

**Struktura e të dhënave:** stiva — Stack (LIFO: i fundit që hyn, i pari që del), ose rekursion.

**Hapat e algoritmit:**
1. Vendoset kulmi fillestar në stive dhe shënohet si i vizituar.
2. Nxirret kulmi nga fillimi i stives dhe përpunohet.
3. Shtohen në stive të gjithë fqinjët e pavizituar.
4. Veprimi përsëritet derisa stiva të zbrazet ose të arrihet kulmi i kërkuar.

**Veti kryesore:** DFS **nuk e garanton** rrugën më të shkurtër, por shfrytëzon më pak memorie dhe zbatohet më thjesht në mënyrë rekursive.

```
Kompleksiteti kohor:  O(V + E)
```

---

### BFS kundrejt DFS — kur përdoret secili?

| Karakteristika | BFS | DFS |
|----------------|-----|-----|
| A e garanton rrugën më të shkurtër? | Po | Jo |
| Shfrytëzimi i memories | Më i lartë (grafe të gjera) | Më i ulët (grafe të thella) |
| Fusha të zbatimit | Rruga më e shkurtër, përshkimi sipas niveleve | Zbulimi i cikleve, renditja topologjike |
| Zbatimi teknik | Radhë | Stive / rekursion |

---

### Zbatimet në jetën reale

**BFS gjen zbatim në:**
- **navigimin GPS** (Google Maps) — për të përcaktuar rrugën me numrin më të vogël të kthesave ose ndalesave;
- **rrjetet sociale** — për sugjerimin e "personave që ndoshta i njihni" (miqtë e miqve);
- **robotikë** — për planifikimin e rrugës mbi një hartë të tipit rrjetë;
- **Web Crawlers** — për zbulimin e faqeve nivel pas niveli;
- **inteligjencën artificiale në lojëra** — për përcaktimin e mundësive të arritshme në lojërat strategjike.

**DFS gjen zbatim në:**
- **kompilatorë** — për zbulimin e varësive ciklike midis moduleve të kodit;
- **bazat e të dhënave** — për identifikimin e cikleve në marrëdhëniet me çelësa të huaj;
- **zgjidhësit e enigmave** — zgjidhjen e enigmave logjike;
- **analizën e rrjeteve** — për të gjetur grupe kulmesh të lidhura fort midis tyre.

**Si inxhinier, këto algoritme do t'i shfrytëzoni për:**
- projektimin e një rrjeti dhe verifikimin nëse të gjithë kulmet janë të arritshme;
- ndërtimin e sistemeve të tipit build system, ku duhet të zgjidhen varësitë midis komponentëve;
- ndërtimin e sistemeve të gjetjes së rrugës në video-lojëra;
- zhvillimin e sistemeve të rekomandimit, të cilat përshkojnë grafet e njohurive për t'i sugjeruar përdoruesit përmbajtje të lidhura me interesat e tij.

## Ushtrimi 1 — Zgjidhja e Labirintit me BFS dhe DFS


````python
import tkinter as tk
import threading
import time
from collections import deque

MAZE_TEXT = """\
###                 #########
#   ###################   # #
# ####          #     # # # #
# ########## ######## # # # #
#                     # # # #
##################### # # # #
#   ##                # #   #
# # ## ### ## ######### # # #
# #    #   ##B#         # # #
# # ## ################ # ###
# # ##             #### # # #
# # ############## ## # # # #
# #             ##    # # # #
# #### ######## ####### # # #
######    #             #   #
A      ######################"""

GAP = 2
COLORS = {
    "wall":      "#808080",
    "space":     "#000000",
    "start":     "#ff4500",
    "goal":      "#32cd32",
    "explored":  "#1e6fb5",   
    "path":      "#ffff00",   
    "frontier":  "#ff8800",   
    "bg":        "#000000",
    "btn_bg":    "#a6fabd",
    "btn_fg":    "#01671e",
    "btn_hover": "#01671e",
    "btn_h_fg":  "#a6fabd",
    "hover":     "#334433",
}

DIRS = [(-1, 0), (1, 0), (0, -1), (0, 1)]


class MazeSolver:
    def __init__(self, root):
        self.root = root
        self.root.title("Maze Solver")
        self.root.configure(bg=COLORS["bg"])

        # FIX C: use a threading.Event instead of a bare bool.
        self.solving_event = threading.Event()

        # FIX G: signal worker threads to abort when the window is closed.
        self.cancelled = threading.Event()

        self.step_mode  = tk.BooleanVar(value=False)
        self.step_event = threading.Event()

        # fullscreen
        self.root.attributes("-fullscreen", True)
        self.root.bind("<Escape>", self._on_escape)
        self.root.bind("<space>",  self._on_space)
        self.root.protocol("WM_DELETE_WINDOW", self._on_escape)
        self.root.update_idletasks()
        scr_w = self.root.winfo_screenwidth()
        scr_h = self.root.winfo_screenheight()

        # parse maze
        rows = MAZE_TEXT.split("\n")
        self.height = len(rows)
        self.width  = max(len(r) for r in rows)
        self.walls  = []
        self.open_cells = set()

        for i, row in enumerate(rows):
            wall_row = []
            for j in range(self.width):
                ch = row[j] if j < len(row) else " "
                if ch == "#":
                    wall_row.append(True)
                else:
                    wall_row.append(False)
                    self.open_cells.add((i, j))
            self.walls.append(wall_row)

        self.start   = None
        self.goal    = None
        self.placing = "start"   # "start" → "goal" → "ready"

        # compute cell size to fill screen
        title_space = 80
        btn_space   = 140
        avail_h = scr_h - title_space - btn_space
        avail_w = scr_w - 60

        cell_from_w = (avail_w - GAP) // self.width  - GAP
        cell_from_h = (avail_h - GAP) // self.height - GAP
        self.cell   = max(8, min(cell_from_w, cell_from_h))

        grid_w = self.width  * (self.cell + GAP) + GAP
        grid_h = self.height * (self.cell + GAP) + GAP

        # title
        font_size = max(18, self.cell * 2 // 3)
        tk.Label(root, text="BFS and DFS maze solving simulation",
                 font=("Helvetica", font_size, "bold"),
                 bg=COLORS["bg"], fg="white").pack(pady=(24, 12))

        # canvas
        self.canvas = tk.Canvas(root, width=grid_w, height=grid_h,
                                bg=COLORS["bg"], highlightthickness=0)
        self.canvas.pack()
        self.canvas.bind("<Button-1>", self._on_click)
        self.canvas.bind("<Motion>",   self._on_hover)

        # controls row
        btn_font = max(16, self.cell // 2)
        ctrl = tk.Frame(root, bg=COLORS["bg"])
        ctrl.pack(pady=(18, 0))

        for algo in ("BFS", "DFS"):
            tk.Button(
                ctrl, text=algo,
                font=("Courier New", btn_font),
                bg=COLORS["btn_bg"], fg=COLORS["btn_fg"],
                activebackground=COLORS["btn_hover"],
                activeforeground=COLORS["btn_h_fg"],
                relief="solid", bd=1, width=12,
                command=lambda a=algo: self.start_solve(a),
            ).pack(side="left", padx=10)

        tk.Checkbutton(
            ctrl, text="Step-by-step (Space)",
            variable=self.step_mode,
            font=("Courier New", max(12, btn_font - 4)),
            bg=COLORS["bg"], fg="#a6fabd",
            selectcolor=COLORS["bg"],
            activebackground=COLORS["bg"],
            activeforeground="#a6fabd",
        ).pack(side="left", padx=(24, 0))

        tk.Button(
            ctrl, text="Reset",
            font=("Courier New", btn_font),
            bg="#ff6666", fg="white",
            activebackground="#cc3333", activeforeground="white",
            relief="solid", bd=1, width=10,
            command=self._reset,
        ).pack(side="left", padx=(24, 0))

        # status bar
        self.status = tk.Label(root, text="",
                               font=("Courier New", max(12, btn_font - 4)),
                               bg=COLORS["bg"], fg="#888888")
        self.status.pack(pady=(10, 0))

        # legend
        legend = tk.Frame(root, bg=COLORS["bg"])
        legend.pack(pady=(4, 0))
        for label, colour in [("Explored", COLORS["explored"]),
                               ("Frontier", COLORS["frontier"]),
                               ("Path",     COLORS["path"])]:
            tk.Label(legend, text="■ ", fg=colour,
                     font=("Helvetica", 12), bg=COLORS["bg"]).pack(side="left")
            tk.Label(legend, text=label + "  ", fg="#888888",
                     font=("Helvetica", 12), bg=COLORS["bg"]).pack(side="left")

        # hint
        tk.Label(root, text="Escape = exit",
                 font=("Helvetica", 10), bg=COLORS["bg"],
                 fg="#333333").pack(side="bottom", pady=(0, 10))

        # draw
        self.cell_ids   = {}
        self.hover_cell = None
        self._draw_maze()
        self._set_status("Click a black cell to place the START point")

    # window / keyboard events

    def _on_escape(self, event=None):
        self.cancelled.set()
        self.step_event.set()
        self.root.destroy()

    def _on_space(self, event):
        self.step_event.set()

    # pixel → grid

    def _pixel_to_cell(self, px, py):
        j = (px - GAP) // (self.cell + GAP)
        i = (py - GAP) // (self.cell + GAP)
        if 0 <= i < self.height and 0 <= j < self.width:
            return i, j
        return None

    # hover highlight

    def _on_hover(self, event):
        if self.placing == "ready" or self.solving_event.is_set():
            return
        cell = self._pixel_to_cell(event.x, event.y)
        if cell == self.hover_cell:
            return
        if self.hover_cell and self.hover_cell in self.cell_ids:
            old = self.hover_cell
            if old != self.start and old != self.goal and old in self.open_cells:
                self.canvas.itemconfig(self.cell_ids[old], fill=COLORS["space"])
        self.hover_cell = cell
        if cell and cell in self.open_cells and cell != self.start and cell != self.goal:
            self.canvas.itemconfig(self.cell_ids[cell], fill=COLORS["hover"])

    # click to place start / goal 

    def _on_click(self, event):
        if self.solving_event.is_set():
            return
        cell = self._pixel_to_cell(event.x, event.y)
        if not cell or cell not in self.open_cells:
            return

        if self.placing == "start":
            if self.start and self.start in self.cell_ids:
                self.canvas.itemconfig(self.cell_ids[self.start], fill=COLORS["space"])
            self.start = cell
            self.canvas.itemconfig(self.cell_ids[cell], fill=COLORS["start"])
            self.placing = "goal"
            self._set_status("Now click a black cell to place the GOAL point")

        elif self.placing == "goal":
            if cell == self.start:
                return
            if self.goal and self.goal in self.cell_ids:
                self.canvas.itemconfig(self.cell_ids[self.goal], fill=COLORS["space"])
            self.goal = cell
            self.canvas.itemconfig(self.cell_ids[cell], fill=COLORS["goal"])
            self.placing = "ready"
            self._set_status("Ready — click BFS or DFS to solve!")

    # reset───

    def _reset(self):
        if self.solving_event.is_set():
            return
        self.start      = None
        self.goal       = None
        self.placing    = "start"
        self.hover_cell = None
        self._draw_maze()
        self._set_status("Click a black cell to place the START point")

    # drawing helpers

    def _cell_rect(self, i, j):
        x1 = GAP + j * (self.cell + GAP)
        y1 = GAP + i * (self.cell + GAP)
        return x1, y1, x1 + self.cell, y1 + self.cell

    def _draw_maze(self):
        self.canvas.delete("all")
        self.cell_ids = {}
        for i in range(self.height):
            for j in range(self.width):
                x1, y1, x2, y2 = self._cell_rect(i, j)
                if (i, j) == self.start:
                    c = COLORS["start"]
                elif (i, j) == self.goal:
                    c = COLORS["goal"]
                elif self.walls[i][j]:
                    c = COLORS["wall"]
                else:
                    c = COLORS["space"]
                rid = self.canvas.create_rectangle(
                    x1, y1, x2, y2, fill=c, outline="", width=0)
                self.cell_ids[(i, j)] = rid

    def _color_cell(self, i, j, color):
        self.canvas.itemconfig(self.cell_ids[(i, j)], fill=color)

    def _set_status(self, text):
        self.status.config(text=text)

    # wait helper──

    def _wait_step(self, explored_n, frontier_n):
        if self.cancelled.is_set():
            return
        if self.step_mode.get():
            self.root.after(0, self._set_status,
                f"Explored: {explored_n}  |  Frontier: {frontier_n}  |  Press Space →")
            self.step_event.clear()
            self.step_event.wait()
        else:
            self.root.after(0, self._set_status,
                f"Explored: {explored_n}  |  Frontier: {frontier_n}")
            time.sleep(0.04)

    # path reconstruction from parent map──────────────────────

    @staticmethod
    def _reconstruct(parent, goal):
        path, cur = [], goal
        while cur is not None:
            path.append(cur)
            cur = parent[cur]
        path.reverse()
        return path

    # FIX E: batch fade-out ──────
    # Collecting all cells to clear and issuing a *single* root.after call avoids
    # flooding the tkinter event queue with thousands of micro-callbacks, which
    # caused visible jank on large mazes in the original implementation.

    def _fade_explored(self, explored, path_set):
        cells_to_clear = [
            c for c in explored
            if c not in path_set and c != self.start and c != self.goal
        ]
        self.root.after(0, lambda: [
            self._color_cell(c[0], c[1], COLORS["space"])
            for c in cells_to_clear
        ])

    # entry point──

    def start_solve(self, algorithm):
        if self.solving_event.is_set():
            return
        if not self.start or not self.goal:
            self._set_status("Place both START and GOAL first!")
            return
        self.solving_event.set()
        self.step_event.clear()
        self._draw_maze()
        # FIX F: dispatch to dedicated BFS / DFS helpers rather than branching
        # inside a single merged _solve method.
        target = self._solve_bfs if algorithm == "BFS" else self._solve_dfs
        threading.Thread(target=target, daemon=True).start()

    # shared finish helper ───────

    def _finish(self, found_path, explored):
        if self.cancelled.is_set():
            self.solving_event.clear()
            return

        if found_path:
            self.root.after(0, self._set_status,
                f"Done!  Path length: {len(found_path)}  |  Explored: {len(explored)}")
            for cell in found_path[1:-1]:
                self.root.after(0, self._color_cell,
                                cell[0], cell[1], COLORS["path"])
        else:
            self.root.after(0, self._set_status,
                f"No path found!  Explored: {len(explored)}")

        time.sleep(0.8)

        # FIX E: single batched after() call instead of one per cell.
        path_set = set(found_path)
        self._fade_explored(explored, path_set)

        self.solving_event.clear()
        self.placing = "ready"

    def _solve_bfs(self):
        parent         = {self.start: None}
        queue          = deque([self.start])
        frontier_cells = {self.start}
        explored       = set()
        found_path     = []

        while queue and not self.cancelled.is_set():
            state = queue.popleft()           # FIFO — correct for BFS
            frontier_cells.discard(state)

            if state in explored:
                continue
            explored.add(state)

            if state != self.start and state != self.goal:
                self.root.after(0, self._color_cell,
                                state[0], state[1], COLORS["explored"])

            found = False
            for di, dj in DIRS:
                ni, nj    = state[0] + di, state[1] + dj
                neighbour = (ni, nj)

                if not (0 <= ni < self.height and 0 <= nj < self.width):
                    continue
                if self.walls[ni][nj]:
                    continue
                if neighbour in explored or neighbour in frontier_cells:
                    continue

                parent[neighbour] = state
                frontier_cells.add(neighbour)
                queue.append(neighbour)

                if neighbour == self.goal:
                    # Early exit: goal found at push time (optimal for BFS).
                    found_path = self._reconstruct(parent, self.goal)
                    found = True
                    break

                self.root.after(0, self._color_cell, ni, nj, COLORS["frontier"])

            if found:
                break
            self._wait_step(len(explored), len(queue))

        self._finish(found_path, explored)

    def _solve_dfs(self):
        parent       = {self.start: None}
        stack        = deque([self.start])
        dfs_on_stack = {self.start}
        explored     = set()
        found_path   = []

        while stack and not self.cancelled.is_set():
            state = stack.pop()               # LIFO — correct for DFS
            dfs_on_stack.discard(state)       # FIX A: release slot for backtrack

            if state in explored:
                continue
            explored.add(state)

            if state != self.start and state != self.goal:
                self.root.after(0, self._color_cell,
                                state[0], state[1], COLORS["explored"])

            found = False
            for di, dj in DIRS:
                ni, nj    = state[0] + di, state[1] + dj
                neighbour = (ni, nj)

                if not (0 <= ni < self.height and 0 <= nj < self.width):
                    continue
                if self.walls[ni][nj]:
                    continue
                if neighbour in explored:
                    continue
                if neighbour in dfs_on_stack:  # FIX A: skip if already stacked
                    continue

                # FIX D: always overwrite — keeps parent consistent with the
                # branch DFS is currently following, not an earlier abandoned one.
                parent[neighbour] = state
                dfs_on_stack.add(neighbour)
                stack.append(neighbour)

                if neighbour == self.goal:
                    found_path = self._reconstruct(parent, self.goal)
                    found = True
                    break

                self.root.after(0, self._color_cell, ni, nj, COLORS["frontier"])

            if found:
                break
            self._wait_step(len(explored), len(stack))

        self._finish(found_path, explored)


if __name__ == "__main__":
    root = tk.Tk()
    MazeSolver(root)
    root.mainloop()
````

## Ushtrimi 2 — Ndërtimi interaktiv i grafit me BFS dhe DFS

**Dallimi kryesor nga Ushtrimi 1:**

| Ushtrimi 1 — Labirinti | Ushtrimi 2 — Ndërtuesi interaktiv |
|------------------------|-----------------------------------|
| Graf i nënkuptuar (rrjetë dydimensionale) | Graf i paraqitur qartë (kulme dhe brinjë) |
| Pa peshë | I drejtuar (brinjët kanë drejtim) |
| Grafi i paracaktuar | Grafin e ndërtoni vetë |
| Vizualizim me ngjyra në rrjetë | Vizualizim me shigjeta midis kulmeve |
---

### Çfarë duhet vëzhguar

Ndërtoni të njëjtin graf dhe më pas ekzekutoni BFS dhe DFS duke u nisur nga i njëjti kulm fillestar. Vini re:

1. **Rendi i zbulimit** — BFS i viziton fillimisht kulmet më afër pikës së nisjes, ndërsa DFS zhytet menjëherë në thellësi.
2. **Rruga e gjetur** — BFS gjen gjithmonë rrugën më të shkurtër deri te destinacioni, ndërsa DFS mund të gjejë një rrugë më të gjatë.
3. **Numri i hapave** — Krahasoni sa hapa i nevojiten secilit algoritëm për të përfunduar.

## Ushtrimi 2 — Ndërtimi Interaktiv i Grafit me BFS dhe DFS


````python
import sys
import pygame
import math
from collections import deque

pygame.init()

info         = pygame.display.Info()
WIDTH, HEIGHT= info.current_w, info.current_h
screen       = pygame.display.set_mode((WIDTH, HEIGHT), pygame.FULLSCREEN)
pygame.display.set_caption("Interactive BFS / DFS Graph Builder")

FONT   = pygame.font.SysFont("Segoe UI", 16)

BG            = (15,  20,  35)
NODE_R        = 20
C_NODE        = (80,  100, 140)
C_NODE_VIS    = (40,  160, 80)
C_NODE_TARGET = (30,  120, 220)
C_EDGE        = (80,  90,  120)
C_EDGE_ACTIVE = (255, 80,  80)
C_PATH        = (255, 160, 30)
UI_BG         = (25,  30,  50)
BTN_BG        = (60,  80,  120)
BTN_HOVER     = (90,  120, 180)
BTN_TEXT      = (255, 255, 255)


def quit_game():
    pygame.quit()
    sys.exit()


# ─────────────────────────────────────────────
# ARROW DRAW
# ─────────────────────────────────────────────
def draw_arrow(surf, color, start, end, width=2, head=12, rad=0.13):
    dx, dy = end[0]-start[0], end[1]-start[1]
    if math.hypot(dx, dy) < 1:
        return
    mx = (start[0]+end[0])/2 + rad*(-dy)
    my = (start[1]+end[1])/2 + rad*(dx)
    pts = [
        (
            (1-t/20)**2*start[0] + 2*(1-t/20)*(t/20)*mx + (t/20)**2*end[0],
            (1-t/20)**2*start[1] + 2*(1-t/20)*(t/20)*my + (t/20)**2*end[1],
        )
        for t in range(21)
    ]
    if len(pts) > 1:
        pygame.draw.lines(surf, color, False,
                          [(int(x), int(y)) for x, y in pts], width)

    def head_draw(tx, ty, adx, ady):
        l = (tx - head*adx + head*0.4*(-ady),
             ty - head*ady + head*0.4*(adx))
        r = (tx - head*adx - head*0.4*(-ady),
             ty - head*ady - head*0.4*(adx))
        pygame.draw.polygon(surf, color, [(tx, ty), l, r])

    for i in range(len(pts)-1, 0, -1):
        px, py = pts[i]
        if math.hypot(px-end[0], py-end[1]) >= NODE_R+4:
            nx, ny   = pts[i-1]
            adx, ady = px-nx, py-ny
            l = math.hypot(adx, ady)
            if l < 0.1:
                break
            head_draw(px, py, adx/l, ady/l)
            break


# ─────────────────────────────────────────────
# BFS — FIFO via deque.popleft()
# ─────────────────────────────────────────────
def bfs(adj, s):
    n       = len(adj)
    marked  = [False] * n
    edge_to = [-1]    * n
    steps   = []

    q        = deque([s])
    marked[s]= True

    print("\n" + "="*60)
    print(f"BFS STARTED  —  source: {s}")
    print("="*60)
    for i, nb in enumerate(adj):
        print(f"  adj[{i}] = {nb}")
    print(f"  Initial queue (FIFO): {list(q)}")

    while q:
        v = q.popleft()
        steps.append(("visit", v, None, marked[:], edge_to[:]))
        print(f"\n  [VISIT] node {v}  |  queue: {list(q)}")

        for w in adj[v]:
            if not marked[w]:
                marked[w]  = True
                edge_to[w] = v
                q.append(w)
                steps.append(("edge", v, w, marked[:], edge_to[:]))
                print(f"    enqueue {v}→{w}  | queue: {list(q)}")
            else:
                steps.append(("skip", v, w, marked[:], edge_to[:]))
                print(f"    skip    {v}→{w}  (visited)")

    print("\n" + "="*60)
    print(f"BFS FINISHED  edge_to: {edge_to}")
    print("="*60 + "\n")
    return steps


# ─────────────────────────────────────────────
# DFS — LIFO via stack
# Push ALL neighbors at once.
# LIFO picks the LAST pushed neighbor first,
# which is the LAST edge you drew from that node.
# ─────────────────────────────────────────────
def dfs(adj, s):
    n       = len(adj)
    marked  = [False] * n
    edge_to = [-1]    * n
    steps   = []
    stack   = [s]

    print("\n" + "="*60)
    print(f"DFS STARTED  —  source: {s}")
    print("="*60)
    for i, nb in enumerate(adj):
        print(f"  adj[{i}] = {nb}")
    print(f"  Initial stack (LIFO): {stack}")

    while stack:
        v = stack.pop()                    # LIFO

        if marked[v]:
            print(f"\n  [POP] {v} already visited — skip")
            continue

        marked[v] = True
        steps.append(("visit", v, None, marked[:], edge_to[:]))
        print(f"\n  [VISIT] node {v}")
        print(f"    adj[{v}] = {adj[v]}")
        print(f"    pushing all unvisited neighbors → LIFO picks LAST one first")

        for w in adj[v]:
            if not marked[w]:
                edge_to[w] = v
                stack.append(w)
                steps.append(("edge", v, w, marked[:], edge_to[:]))
                print(f"    push {w}  | stack now: {stack}")

    print("\n" + "="*60)
    print(f"DFS FINISHED  edge_to: {edge_to}")
    print("="*60 + "\n")
    return steps


# ─────────────────────────────────────────────
# APP
# ─────────────────────────────────────────────
class App:
    def __init__(self):
        self.nodes   = []
        self.edges   = []     # LIST preserves draw order exactly
        self.history = []

        self.mode        = "build"
        self.edge_start  = None
        self.start       = None
        self.target      = None

        self.algo         = "BFS"
        self.steps        = []
        self.step_i       = 0
        self.control_mode = "play"
        self.play         = False
        self.speed        = 3
        self.timer        = 0

        self.btn_run   = pygame.Rect(WIDTH-380, HEIGHT-180, 300, 45)
        self.btn_reset = pygame.Rect(WIDTH-380, HEIGHT-120, 300, 45)

    def add_node(self, pos):
        nid = len(self.nodes)
        self.nodes.append(pos)
        self.history.append(("node", nid))
        print(f"\n[NODE ADDED]  id={nid}")

    def _edge_exists(self, u, v):
        return any(eu == u and ev == v for eu, ev in self.edges)

    def add_edge(self, u, v):
        if u == v:
            return
        if self._edge_exists(u, v):
            print(f"\n[EDGE SKIPPED]  {u}→{v}  duplicate")
            return
        self.edges.append((u, v))
        self.history.append(("edge", u, v))
        print(f"\n[EDGE ADDED]  {u}→{v}  order: {self.edges}")

    def undo(self):
        if not self.history:
            return
        last = self.history.pop()
        if last[0] == "node":
            idx = last[1]
            self.nodes.pop(idx)
            self.edges = [(u, v) for u, v in self.edges
                          if u != idx and v != idx]
        elif last[0] == "edge":
            self.edges.remove((last[1], last[2]))

    def clear(self):
        self.nodes.clear()
        self.edges.clear()
        self.history.clear()
        self.start  = None
        self.target = None
        self.reset_algo()
        print("\n[CLEAR]")

    def build_adj(self):
        adj = [[] for _ in self.nodes]
        for u, v in self.edges:
            adj[u].append(v)
        print("\n" + "-"*50)
        print("ADJACENCY LIST (matches draw order):")
        for i, nb in enumerate(adj):
            print(f"  adj[{i}] = {nb}")
        print("-"*50)
        return adj

    def reset_algo(self):
        self.step_i = 0
        self.play   = False
        self.timer  = 0
        self.steps  = []

    def run_algo(self):
        if self.start is None:
            return
        adj = self.build_adj()
        self.steps  = bfs(adj, self.start) if self.algo == "BFS" \
                      else dfs(adj, self.start)
        self.step_i = 0
        self.play   = False
        self.timer  = 0

    def next_step(self):
        if not self.steps:
            return
        if self.step_i < len(self.steps) - 1:
            self.step_i += 1
            self._print_step()
        else:
            print("\n[END OF STEPS]")

    def prev_step(self):
        if not self.steps:
            return
        if self.step_i > 0:
            self.step_i -= 1
            self._print_step()
        else:
            print("\n[ALREADY AT STEP 0]")

    def _print_step(self):
        kind, v, w, marked, edge_to = self.steps[self.step_i]
        visited = [i for i, m in enumerate(marked) if m]
        print(f"\n[STEP {self.step_i}/{len(self.steps)-1}]"
              f"  {kind:5s}  node={v}  neighbor={w}  visited={visited}")

    def get_path(self):
        if self.target is None or not self.steps:
            return []
        edge_to = self.steps[self.step_i][4]
        path, x = [], self.target
        seen = set()
        while x != -1:
            if x in seen:
                break
            seen.add(x)
            path.append(x)
            x = edge_to[x]
        path.reverse()
        if path and path[0] != self.start:
            return []
        return path

    def handle_click(self, pos):
        if self.btn_run.collidepoint(pos):
            self.run_algo()
            return
        if self.btn_reset.collidepoint(pos):
            self.reset_algo()
            return
        for i, (x, y) in enumerate(self.nodes):
            if math.hypot(pos[0]-x, pos[1]-y) < NODE_R:
                self.handle_node_click(i)
                return
        if self.mode == "build":
            self.add_node(pos)

    def handle_node_click(self, i):
        if self.mode == "build":
            if self.edge_start is None:
                self.edge_start = i
            else:
                self.add_edge(self.edge_start, i)
                self.edge_start = None
        elif self.mode == "start":
            self.start = i
            print(f"\n[START] node {i}")
            self.run_algo()
        elif self.mode == "target":
            self.target = i
            print(f"\n[TARGET] node {i}")

    def draw_button(self, rect, text):
        mouse = pygame.mouse.get_pos()
        color = BTN_HOVER if rect.collidepoint(mouse) else BTN_BG
        pygame.draw.rect(screen, color, rect, border_radius=8)
        pygame.draw.rect(screen, (120, 150, 220), rect, 2, border_radius=8)
        lbl = FONT.render(text, True, BTN_TEXT)
        screen.blit(lbl, lbl.get_rect(center=rect.center))

    def draw(self):
        screen.fill(BG)

        for u, v in self.edges:
            draw_arrow(screen, C_EDGE, self.nodes[u], self.nodes[v], 2)

        if self.steps:
            kind, v, w, marked, edge_to = self.steps[self.step_i]
            if w is not None:
                draw_arrow(screen, C_EDGE_ACTIVE,
                           self.nodes[v], self.nodes[w], 4)

        path = self.get_path()
        for i in range(len(path)-1):
            draw_arrow(screen, C_PATH,
                       self.nodes[path[i]], self.nodes[path[i+1]], 4)

        if self.edge_start is not None:
            mx, my = pygame.mouse.get_pos()
            draw_arrow(screen, (150, 150, 255),
                       self.nodes[self.edge_start], (mx, my), 2)

        for i, (x, y) in enumerate(self.nodes):
            col = C_NODE
            if self.steps and self.steps[self.step_i][3][i]:
                col = C_NODE_VIS
            if i == self.target:
                col = C_NODE_TARGET      # blue
            if i == self.start:
                col = (200, 60, 60)      # red — always on top
            pygame.draw.circle(screen, col, (x, y), NODE_R)
            lbl = FONT.render(str(i), True, (255, 255, 255))
            screen.blit(lbl, lbl.get_rect(center=(x, y)))

        pygame.draw.rect(screen, UI_BG, (WIDTH-400, 0, 400, HEIGHT))
        y = 10
        def t(s):
            nonlocal y
            screen.blit(FONT.render(s, True, (200, 220, 255)), (WIDTH-390, y))
            y += 22

        t(f"Mode      : {self.mode}")
        t(f"Algorithm : {self.algo}")
        t(f"Control   : {self.control_mode}")
        t(f"Step      : {self.step_i} / {max(0, len(self.steps)-1)}")
        t(f"Path      : {' → '.join(map(str, path)) if path else '—'}")
        t("")
        t("── Colors ────────────────────")
        t("Red    : start node")
        t("Blue   : target node")
        t("Green  : visited node")
        t("Orange : path found")
        t("")
        t("── Controls ──────────────────")
        t("B : build mode")
        t("S : set start node")
        t("T : set target node")
        t("D : use DFS")
        t("F : use BFS")
        t("M : toggle play / step")
        t("SPACE : play / pause")
        t("N : next step")
        t("P : prev step")
        t("Z : undo")
        t("C : clear graph")
        t("ESC : exit")

        self.draw_button(self.btn_run,   "Run Algorithm")
        self.draw_button(self.btn_reset, "Reset Steps")
        pygame.display.flip()

    def run(self):
        clock = pygame.time.Clock()
        print("\n" + "="*60)
        print("  BFS / DFS Visualizer  —  press ESC to exit")
        print("="*60)

        while True:
            dt = clock.tick(60) / 1000

            if self.control_mode == "play" and self.play and self.steps:
                self.timer += dt
                if self.timer > 1 / self.speed:
                    self.timer = 0
                    if self.step_i < len(self.steps) - 1:
                        self.step_i += 1
                    else:
                        self.play = False
                        print("\n[AUTO]  paused at last step")

            for e in pygame.event.get():

                if e.type == pygame.QUIT:
                    quit_game()

                if e.type == pygame.KEYDOWN:
                    if e.key == pygame.K_ESCAPE:
                        quit_game()                        # clean exit
                    elif e.key == pygame.K_m:
                        self.control_mode = (
                            "step" if self.control_mode == "play" else "play")
                    elif e.key == pygame.K_SPACE:
                        if self.control_mode == "play":
                            self.play = not self.play
                    elif e.key == pygame.K_n:
                        if self.control_mode == "step":
                            self.next_step()
                    elif e.key == pygame.K_p:
                        if self.control_mode == "step":
                            self.prev_step()
                    elif e.key == pygame.K_b:
                        self.mode = "build"
                    elif e.key == pygame.K_s:
                        self.mode = "start"
                    elif e.key == pygame.K_t:
                        self.mode = "target"
                    elif e.key == pygame.K_d:
                        self.algo = "DFS"
                        print("\n[ALGO]  DFS")
                    elif e.key == pygame.K_f:
                        self.algo = "BFS"
                        print("\n[ALGO]  BFS")
                    elif e.key == pygame.K_z:
                        self.undo()
                    elif e.key == pygame.K_c:
                        self.clear()

                if e.type == pygame.MOUSEBUTTONDOWN:
                    self.handle_click(e.pos)

            self.draw()


if __name__ == "__main__":
    App().run()
````

## Ushtrimi 3 — Pema Perfshirese Minimum: Kruskal dhe Prim

### Çfarë është një Pemë Perfshirese?

Duke pasur një **graf të lidhur, pa peshë, të pa drejtuar** me N kulme, një **Pemë Perfshirese** është nëngrafi që:
- Lidh **të gjitha kulmet** (mbulon grafin)
- Përmban saktësisht **N − 1 brinjë** (pa cikle)
- Është një **pemë** (e lidhur dhe acilike)

Një **Pemë Perfshirese Minimum (MST)** është pema perfshirese ku **shuma totale e peshëve të brinjëve është minimale**.

> - **Analogji:** Imagino se je planifikues qyteti dhe duhet të lidhësh 5 lagje me rrugë. Dëshiron që çdo lagje të jetë e arritshme nga çdo tjetër, por dëshiron të ndërtosh sa më pak kilometre rrugë. MST jep zgjidhjen optimale!

---

### Algoritmi i Kruskalit (Bazuar në brinjë / Greedy)

**Ideja kryesore:** Renditi të gjitha brinjët sipas peshës (greedy) shto brinjën me peshë më të vogël që nuk krijon cikël.

**Hapat e algoritmit:**
1. Renditi të gjitha brinjët sipas peshës në rend rritës.
2. Zgjidh brinjën më të vogël. Nëse lidh dy komponente të ndryshme (pa cikël) → shtohet në MST.
3. Përsërit derisa MST të ketë N − 1 brinjë.

**Struktura kryesore:** **Bashkim-Gjej (DSU / Union-Find)** — kontrollon me efikasitet nëse dy nyje janë në të njëjtin komponent.

**Kruskali funksionon më mirë** kur grafi është **i rrallë** (pak brinjë).

---

### Algoritmi i Prim (Bazuar në kulme / Greedy)

**Ideja kryesore:** Rrit MST-n një kulm në të njëjtën kohë. Gjithmonë zgjidhni brinjën me peshë më të vogël që lidh pemën aktuale me një kulm të ri.

**Hapat e algoritmit:**
1. Fillo nga çdo kulm. Shtohet në MST.
2. Shiko të gjitha brinjët që largohen nga MST aktual.
3. Zgjidh brinjën me peshën minimale që shkon te një kulm i pa vizituar.
4. Shto atë kulm dhe brinjë në MST.
5. Përsërit derisa të gjitha kulmet të jenë të përfshirë.

**Struktura kryesore:** **Radhë me Prioritet (Min-Heap)** — nxjerr me efikasitet brinjën me peshë minimale.

**Primi funksionon më mirë** kur grafi është **i dendur** (shumë brinjë).

---

### Kruskal kundrejt Prim — Krahasimi

| Karakteristika | Kruskal | Prim |
|----------------|---------|------|
| Qasja | Bazuar në brinjë | Bazuar në kulme |
| Punon me grafe jo të lidhur? | Po | Kërkon graf të lidhur |
| Më i mirë për | Grafe të rralla | Grafe të dendura |
| Struktura kryesore | Union-Find (DSU) | Radhë me Prioritet |
| Rezultati | E njëjta peshë MST | E njëjta peshë MST |

---

### Zbatimet në Botën Reale të MST

- **Projektimi i Rrjetit Elektrik** — lidhja e qyteteve me rrjetin elektrik me kabllo minimale
- **Telekomunikacioni** — shtrirja e kabllove me fibër optike midis qendrave të të dhënave me kosto minimale
- **Planifikimi i Rrjetit Rrugor** — rrjeti rrugor me kosto minimale që lidh të gjitha vendbanimet
- **Paraqitja e Qarkut të Integruar** — minimizimi i gjatësisë së telit në projektimin e çipeve VLSI
- **Bioinformatika** — ndërtimi i pemëve filogjenetike (pemët evolucionare të llojeve)
- **Robotikë** — planifikimi i rrugës së eksplorimit me kosto minimale për robot
- **Zhvillimi i Lojërave** — gjenerimi i relievit realist rastësor dhe hartave të dungeon-it

**Si inxhinier do të përdorësh MST kur:**
- Planifikon infrastrukturë që minimizon koston e ndërtimit
- Projekton rrjete komunikimi me eliminimin e tepricës
- Grupon pikat e të dhënave në mësim makine (single-linkage clustering)
- Ndërton protokolle rrugëzimi në rrjetet kompjuterike

---

### Çfarë bën ky ushtrim

Vizualizuesi më poshtë të lejon të:
1. Kalon në modalitetin **Sandbox** dhe **klikon** për të shtuar nyje, pastaj **klikon dy nyje** për të shtuar brinjë (pesha caktohet automatikisht).
2. Shtyp **Kruskal** ose **Prim** për të parë MST-n duke u ndërtuar hap pas hapi.
3. Përdor **Next / Prev** për të rishikuar çdo hap, ose **Auto** për luajtje të vazhdueshme.

- **Brinjët cian** = të pranuara në MST
- **Brinjët e kuqe** = të refuzuara (do të krijonin cikël — vetëm Kruskal)


## Ushtrimi 3 — Pema Përfshirëse Minimum: Kruskal dhe Prim


````python
import tkinter as tk
import random
import math

# =========================
# DATA STRUCTURES
# =========================

class Node:
    def __init__(self, nid, x, y):
        self.id = nid
        self.x = x
        self.y = y


class Edge:
    def __init__(self, u, v, w):
        self.u = u
        self.v = v
        self.w = w
        self.highlight = None


# =========================
# DSU
# =========================

class DSU:
    def __init__(self, nodes):
        self.parent = {n: n for n in nodes}
        self.rank = {n: 0 for n in nodes}

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, a, b):
        ra, rb = self.find(a), self.find(b)
        if ra == rb:
            return False
        if self.rank[ra] < self.rank[rb]:
            self.parent[ra] = rb
        elif self.rank[ra] > self.rank[rb]:
            self.parent[rb] = ra
        else:
            self.parent[rb] = ra
            self.rank[ra] += 1
        return True


# =========================
# MAIN APP
# =========================

class MSTApp:
    def __init__(self, root):
        self.root = root
        self.root.title("MST Visualizer")

        self.canvas = tk.Canvas(root, bg="#111")
        self.canvas.pack(fill="both", expand=True)

        self.nodes = []
        self.edges = []

        self.mode = "menu"
        self.selected_node = None

        # Step system
        self.step_index = 0
        self.execution_mode = "manual"
        self.current_algorithm = None

        self.build_ui()
        self.bind_events()

    # =========================
    # UI
    # =========================

    def build_ui(self):
        frame = tk.Frame(self.root, bg="#222")
        frame.place(x=10, y=10)

        tk.Button(frame, text="Prim", command=self.start_prim).pack(side="left")
        tk.Button(frame, text="Kruskal", command=self.start_kruskal).pack(side="left")
        tk.Button(frame, text="Sandbox", command=self.start_sandbox).pack(side="left")

        tk.Button(frame, text="Next", command=self.next_step).pack(side="left")
        tk.Button(frame, text="Prev", command=self.prev_step).pack(side="left")

        tk.Button(frame, text="Auto", command=self.toggle_auto).pack(side="left")
        tk.Button(frame, text="Reset", command=self.reset).pack(side="left")

        # Still allow exit via button
        tk.Button(frame, text="Exit", command=self.root.destroy).pack(side="left")

        self.status = tk.Label(self.root, fg="white", bg="#111")
        self.status.place(x=10, y=50)

    def bind_events(self):
        self.canvas.bind("<Button-1>", self.on_click)
        self.canvas.bind("<B1-Motion>", self.drag_node)

        self.root.bind("n", lambda e: self.next_step())
        self.root.bind("p", lambda e: self.prev_step())
        self.root.bind("<space>", lambda e: self.toggle_auto())

        # ESC to exit (important since no window controls)
        self.root.bind("<Escape>", lambda e: self.root.destroy())

    # =========================
    # GRAPH
    # =========================

    def add_node(self, x, y):
        self.nodes.append(Node(len(self.nodes), x, y))
        self.draw()

    def get_node(self, x, y, r=15):
        for n in self.nodes:
            if math.hypot(n.x - x, n.y - y) < r:
                return n
        return None

    def add_edge(self, u, v):
        if u.id != v.id:
            self.edges.append(Edge(u.id, v.id, random.randint(1, 20)))
        self.draw()

    # =========================
    # SANDBOX
    # =========================

    def start_sandbox(self):
        self.mode = "sandbox"

    def on_click(self, event):
        if self.mode != "sandbox":
            return

        n = self.get_node(event.x, event.y)

        if not n:
            self.add_node(event.x, event.y)
            return

        if not self.selected_node:
            self.selected_node = n
        else:
            self.add_edge(self.selected_node, n)
            self.selected_node = None

    def drag_node(self, event):
        if self.mode != "sandbox":
            return
        n = self.get_node(event.x, event.y)
        if n:
            n.x, n.y = event.x, event.y
            self.draw()

    # =========================
    # RESET
    # =========================

    def reset(self):
        self.nodes = []
        self.edges = []
        self.step_index = 0
        self.current_algorithm = None
        self.execution_mode = "manual"
        self.draw()

    # =========================
    # STEP CONTROL
    # =========================

    def next_step(self):
        self.step_index += 1
        self.replay()

    def prev_step(self):
        if self.step_index > 0:
            self.step_index -= 1
            self.replay()

    # =========================
    # AUTO MODE
    # =========================

    def toggle_auto(self):
        self.execution_mode = "auto" if self.execution_mode == "manual" else "manual"
        if self.execution_mode == "auto":
            self.auto_loop()

    def auto_loop(self):
        if self.execution_mode == "auto":
            self.next_step()
            self.root.after(400, self.auto_loop)

    # =========================
    # ALGORITHMS
    # =========================

    def start_prim(self):
        self.current_algorithm = "prim"
        self.step_index = 0
        self.replay()

    def start_kruskal(self):
        self.current_algorithm = "kruskal"
        self.step_index = 0
        self.replay()

    def replay(self):
        self.clear_highlights()

        if self.current_algorithm == "prim":
            self.replay_prim(self.step_index)
        elif self.current_algorithm == "kruskal":
            self.replay_kruskal(self.step_index)

        self.draw()
        self.update_status()

    def replay_prim(self, steps):
        if not self.nodes:
            return

        visited = {0}

        for _ in range(steps):
            best = None
            for e in self.edges:
                if (e.u in visited) ^ (e.v in visited):
                    if best is None or e.w < best.w:
                        best = e

            if best:
                best.highlight = "mst"
                visited.add(best.u)
                visited.add(best.v)

    def replay_kruskal(self, steps):
        dsu = DSU(range(len(self.nodes)))
        sorted_edges = sorted(self.edges, key=lambda e: e.w)

        for i in range(min(steps, len(sorted_edges))):
            e = sorted_edges[i]
            if dsu.find(e.u) != dsu.find(e.v):
                dsu.union(e.u, e.v)
                e.highlight = "mst"
            else:
                e.highlight = "reject"

    # =========================
    # DRAW
    # =========================

    def clear_highlights(self):
        for e in self.edges:
            e.highlight = None

    def draw(self):
        self.canvas.delete("all")

        for e in self.edges:
            u = self.nodes[e.u]
            v = self.nodes[e.v]

            color = "#555"
            if e.highlight == "mst":
                color = "#00f5ff"
            elif e.highlight == "reject":
                color = "red"

            self.canvas.create_line(u.x, u.y, v.x, v.y, fill=color, width=2)
            self.canvas.create_text((u.x+v.x)/2, (u.y+v.y)/2,
                                    text=str(e.w), fill="white")

        for n in self.nodes:
            self.canvas.create_oval(n.x-15, n.y-15, n.x+15, n.y+15,
                                    fill="#2ecc71")
            self.canvas.create_text(n.x, n.y, text=n.id, fill="black")

    # =========================
    # STATUS
    # =========================

    def update_status(self):
        self.status.config(
            text=f"{self.current_algorithm} | Step {self.step_index} | Mode: {self.execution_mode}"
        )


# =========================
# RUN
# =========================

if __name__ == "__main__":
    root = tk.Tk()
    app = MSTApp(root)
    root.mainloop()
````

## Detyrë për Shtëpi — Problemi 3: Pema Përfshirëse Minimum me Algoritmin e Kruskalit

**Konteksti teorik.**  
Algoritmi i Kruskalit është një algoritëm greedy për gjetjen e Pemës Përfshirëse Minimale (MST) të grafit me peshë të pa drejtuar G = (V, E, w).  
Ai ripërdor strukturën **Bashkim-Gjej (Union-Find / DSU)** për të kontrolluar nëse dy kulme i përkasin të njëjtit komponent.

**Grafi i dhënë.**  
G₃ me peshat e brinjëve:

| Brinja | Pesha |
|--------|-------|
| 0 — 1 | 4 |
| 0 — 2 | 3 |
| 1 — 2 | 1 |
| 1 — 3 | 2 |
| 2 — 3 | 4 |
| 3 — 4 | 2 |
| 2 — 4 | 5 |

**Detyra.**

1. Renditni brinjët dhe gjurmoni Kruskalin me dorë. Shkruani MST T* dhe peshën w(T*).
2. Plotësoni strukturen e kodit duke ripërdorur klasën `DSU` nga Ushtrimi 3 i laboratorit.
3. Verifikoni: numri i brinjëve të MST duhet të jetë |V| − 1 = 4.
4. **Pyetje konceptuale:** Nëse dy brinjë kanë peshë të barabartë, a është MST gjithmonë unike? Jepni shembull.


````python
# ================================================================
# Detyrë për Shtëpi — Problemi 3
# Ripërdorni klasën DSU nga Ushtrimi 3 i laboratorit.
# ================================================================

# --- Klasa DSU (e njëjtë me Ushtrimin 3) — mos e ndrysho ---
class DSU:
    def __init__(self, kulmet):
        self.parent = {v: v for v in kulmet}
        self.rank   = {v: 0 for v in kulmet}

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x]) 
        return self.parent[x]

    def union(self, a, b):
        ra, rb = self.find(a), self.find(b)
        if ra == rb:
            return False          # tashmë në të njëjtin komponent
        if self.rank[ra] < self.rank[rb]: self.parent[ra] = rb
        elif self.rank[ra] > self.rank[rb]: self.parent[rb] = ra
        else: self.parent[rb] = ra; self.rank[ra] += 1
        return True


# --- Grafi G3 si listë brinjësh (u, v, w) ---
brinje_G3 = [
    (0, 1, 4),
    (0, 2, 3),
    (1, 2, 1),
    (1, 3, 2),
    (2, 3, 4),
    (3, 4, 2),
    (2, 4, 5),
]

def kruskal(brinje, kulmet):
    """
    Algoritmi i Kruskalit.
    brinje : listë e çifteve (u, v, w)
    kulmet : listë e kulmeve V
    Kthon (MST, peshe_totale) ku MST eshte lista e brinjeve te pranuara.
    """
    # Hapi 1: Rendit brinjët sipas peshës në rend rritës
    # PLOTESO: plotësoni argumentin e sorted()
    brinje_renditura = sorted(brinje, key=lambda b: ___)

    dsu = DSU(kulmet)
    mst          = []
    peshe_totale = 0

    for u, v, w in brinje_renditura:
        # Hapi 2: Nëse u dhe v janë në komponente të ndryshme, pranohet brinja
        if dsu.union(u, v):          # union kthen True nëse u bashkuan
            mst.append((u, v, w))
            # PLOTESO: shtoni peshën w te peshe_totale
            peshe_totale += ___
            print(f"  Pranohet brinja ({u}, {v})  w={w}  =>  MST deri tani: {len(mst)} brinje")
        else:
            print(f"  Refuzohet brinja ({u}, {v})  w={w}  (do formonte cikël)")

    return mst, peshe_totale


# --- Ekzekutimi ---
kulmet_G3 = list(range(5))   # V = {0, 1, 2, 3, 4}

print("Gjurmimi i Kruskalit mbi G3:")
mst, w_total = kruskal(brinje_G3, kulmet_G3)

print(f"\nPema Përfshirëse Minimum T*:")
for u, v, w in mst:
    print(f"  ({u}, {v})  pesha={w}")

print(f"\nPesha totale w(T*) = {w_total}")
print(f"Numri i brinjeve te MST = {len(mst)}  (pritet {len(kulmet_G3) - 1})")
````
